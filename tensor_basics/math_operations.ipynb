```python
import tensorflow as tf
```

```python
x_abs=tf.constant([-2.5,3.25])
tf.abs(x_abs)
tf.abs(tf.constant(-0.2))
x_abs_complex=tf.constant([-2.45+4.75j])
tf.abs(x_abs_complex)
```

```python
tf.sqrt((-2.5)**2 + 4.75*2)
```

```python
x1=tf.constant([5,3,6,6,4,6],dtype=tf.int32)
x2=tf.constant([7,6,2,6,7,11],dtype=tf.int32)
print(tf.add(x1,x2))
print(tf.multiply(x1,x2))
print(tf.divide(x1,x2))
print(tf.subtract(x1,x2))
```

```python
x3=tf.constant([5,3,6,6,4,6],dtype=tf.float32)
x4=tf.constant([7,6,2,6,0,11],dtype=tf.float32)
print(tf.math.divide_no_nan(x3,4))
```

```python
x5=tf.constant([5,3,6,6,4,6],dtype=tf.float32)
x6=tf.constant([7],dtype=tf.float32)
print(tf.math.add(x5,x6))
```

```python
y1=tf.constant([5,3,6,6,4,6],dtype=tf.float32)
y2=tf.constant([[7],[5],[3]],dtype=tf.float32)
print(y1.shape)
print(y2.shape)
print(tf.math.multiply(y1,y1))
```

```python
z=tf.constant([200,120,130,300,60])
print(tf.math.argmax(z))
print(tf.math.argmin(z))
```

```python
j=tf.constant([[2,20,30,3,6],
              [3,11,16,1,8],
              [14,45,23,5,27]])
print(tf.math.argmax(j,0))
print(tf.math.argmax(j,1))
```

```python
a=tf.constant([[2,2],[3,3]])
b=tf.constant([[3,1],[1,4]])
tf.pow(a,b)
```

```python
d=tf.constant([[1,2,0],
               [3,5,-1],
               [1,5,6],
               [2,3,8]])
print(tf.math.reduce_sum(d,axis=None,keepdims=False,name=None))
print(tf.math.reduce_sum(d,axis=0,keepdims=False,name=None))
print(tf.math.reduce_max(d,axis=None,keepdims=False,name=None))
print(tf.math.reduce_min(d,axis=None,keepdims=False,name=None))
```

```python
s=tf.constant([[1,2,0],
               [3,5,-1],
               [1,5,6],
               [2,3,8]])
tf.math.top_k(s,k=1)
```
