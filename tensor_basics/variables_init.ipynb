```python
import tensorflow as tf
```

```python
tensor_0d=tf.constant(4)
print(tensor_0d)
```

```python
tensor_1d=tf.constant([2,0,-3])
print(tensor_1d)
```

```python
tensor_oneD=tf.constant([2,0.1,3])
print(tensor_oneD)
```

```python
tensor_2d=tf.constant([[1,2,0],
                       [3,5,-1],
                       [1,5,6],
                       [2,3,8]])
print(tensor_2d)
```

```python
tensor_3d=tf.constant([[[1,2,0],
                        [3,5,1]],
                        [[10,2,0],
                         [1,0,2]],
                       [[5,8,0],
                        [2,7,0]],
                       [[5,8,0],
                        [2,7,0]],
                       [[2,1,9],
                        [4,-3,32]]])
print(tensor_3d)
```

```python
tensor_oneD=tf.constant([2,0.2,22],dtype=tf.float32)
casted_tensor=tf.cast(tensor_oneD,dtype=tf.int32)
print(tensor_oneD)
print(casted_tensor)
```
